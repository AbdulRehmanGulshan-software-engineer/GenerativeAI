```python
import random

class fakeLLM:
    
    # constructor
    def __init__(self):
        print('LLM created')
        
    def predict(self, prompt):
        
        # for now generate dummy response here
        response_list = [
            'Islamabad is the capital of Pakistan',
            'PSL is a cricket League',
            'AI stands for Artificial Intelligence'
        ]
        
        return {'response': random.choice(response_list)}
```

```python
llm = fakeLLM()
```

```python
llm.predict('what is the capital of Pakistan?')
```

### Making Fake Prompt Template

```python
class fakePromptTemplate:
    
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables
        
    def format(self,inpuut_dic):
        return self.template.format(**inpuut_dic)
```

```python
template = fakePromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length','topic']
)
```

```python
prompt = template.format({'length':'short','topic':'Pakistan'})
```

```python
llm = fakeLLM()
```

```python
llm.predict(prompt)
```

### Lets make a fake LLM Chain
#### its work is to connect Prompt Template and LLM Component

```python
class fakeLLMChain:
    def __init__(self, llm, prompt):
        self.llm = llm
        self.prompt = prompt
    
    def run(self, input_data):
        
        final_prompt = self.prompt.format(input_data)
        result = self.llm.predict(input_data)
        
        # abstract string from result
        return result['response']
```

```python
template = fakePromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length','topic']
)
```

```python
llm = fakeLLM()
```

```python
chain = fakeLLMChain(llm,template)
```

```python
chain.run({'length':'short', 'topic':'Saudi Arabia'})
```
