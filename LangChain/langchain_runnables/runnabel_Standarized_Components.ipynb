```python
from abc import ABC, abstractmethod
```

```python
class Runnable(ABC):

    @abstractmethod
    def invoke(self, input_data):
        pass
```

#### Fake LLM Component Class

```python
import random

class fakeLLM(Runnable):
    
    # constructor
    def __init__(self):
        print('LLM created')
        
    # making invoke method due to abstract class condition for it
    def invoke(self, prompt):
        response_list = [
            'Islamabad is the capital of Pakistan',
            'PSL is a cricket League',
            'AI stands for Artificial Intelligence'
        ]                
        return {'response': random.choice(response_list)}
                
        
    def predict(self, prompt):
        # for now generate dummy response here
        response_list = [
            'Islamabad is the capital of Pakistan',
            'PSL is a cricket League',
            'AI stands for Artificial Intelligence'
        ]
        
        return {'response': random.choice(response_list)}
```

```python
llm = fakeLLM()
```

#### Fake PromptTemplate Class

```python
class fakePromptTemplate:
    
    def __init__(self,template,input_variables):
        self.template = template
        self.input_variables = input_variables
    
    # making invoke method due to abstract class condition for it
    def invoke(self, input_dict):
        return self.template.format(**input_dict)
        
    def format(self,input_dict):
        return self.template.format(**input_dict)
```

##### Till now we have standarized our components , now we will compose them into chains

Creating Class RunnableConnector , purpose is to connect 2 or many components together

```python
class fakeStrOutputParser(Runnable):
    def __init__(self):
        pass
    
    def invoke(self,input_data):
        return input_data['response']
```

```python
class RunnableConnector(Runnable):
    
    def __init__(self,runnable_list):
        self.runnable_list = runnable_list
        
    def invoke(self,input_data):
        # loop through runnable list
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
    
        return input_data
```

```python
template = fakePromptTemplate(
    template='Write a {length} poem about {topic}',
    input_variables=['length','topic']
)
```

```python
llm = fakeLLM()
```

```python
parser = fakeStrOutputParser()
```

```python
chain = RunnableConnector([template,llm,parser])
```

```python
chain.invoke({'length':'Long', 'topic':'Nipal'})
```
